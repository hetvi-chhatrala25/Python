```python
import numpy as np
```

```python
class DataAnalytics:
    
    def __init__(self):
        self.array = np.array([])

    @staticmethod
    def ShowWelcome():
        print("Welcome to the NumPy Analyzer!")
        
    @classmethod
    def ProjectInfo(cls):
        print("Project: NumPy Analyzer")
        print("Class:", cls.__name__)

    def __GetSecondArray(self, shape):
        total = int(np.prod(shape))

        while True:
            values = input(
                f"Enter the elements of another array to combine ({total} elements separated by space): "
            ).split()

            if len(values) != total:
                print(f"Please enter exactly {total} elements.")
                continue

            try:
                values = [float(x) for x in values]

                if all(x.is_integer() for x in values):
                    values = [int(x) for x in values]

                return np.array(values).reshape(shape)

            except ValueError:
                print("Invalid input. Please enter numbers only.")

    def GetInteger(self, message):
        while True:
            value = input(message).strip()

            if value == "":
                print("Input cannot be empty. Please enter a number.")
                continue

            try:
                return int(value)

            except ValueError:
                print("Invalid input. Please enter a number.")

    def GetArrayValues(self, total):
        while True:
            values = input(
                f"Enter {total} elements for the array separated by space: "
            ).split()

            if len(values) != total:
                print(f"Please enter exactly {total} elements.")
                continue

            try:
                values = [float(x) for x in values]

                if all(x.is_integer() for x in values):
                    values = [int(x) for x in values]

                return values

            except ValueError:
                print("Invalid input. Please enter numbers only.")

    def CheckArray(self):
        if self.array.size == 0:
            print("No array has been created yet.")
            return False
        return True

    def CreateArray(self):

        print("\nArray Creation:")
        print("\nSelect the type of array to create:")
        print("\n1. 1D Array")
        print("\n2. 2D Array")
        print("\n3. 3D Array")

        choice = self.GetInteger("\nEnter your choice: ")

        if choice == 1:

            n = self.GetInteger(
                "Enter the number of elements: "
            )

            if n <= 0:
                print("Number of elements must be greater than 0.")
                return

            values = self.GetArrayValues(n)

            self.array = np.array(values)

        elif choice == 2:

            rows = self.GetInteger(
                "Enter the number of rows: "
            )

            columns = self.GetInteger(
                "Enter the number of columns: "
            )

            if rows <= 0 or columns <= 0:
                print("Rows and columns must be greater than 0.")
                return

            total = rows * columns

            values = self.GetArrayValues(total)

            self.array = np.array(values).reshape(
                rows,
                columns
            )

        elif choice == 3:

            layers = self.GetInteger(
                "Enter the number of layers: "
            )

            rows = self.GetInteger(
                "Enter the number of rows: "
            )

            columns = self.GetInteger(
                "Enter the number of columns: "
            )

            if layers <= 0 or rows <= 0 or columns <= 0:
                print("All dimensions must be greater than 0.")
                return

            total = layers * rows * columns

            values = self.GetArrayValues(total)

            self.array = np.array(values).reshape(
                layers,
                rows,
                columns
            )

        else:
            print("Invalid choice.")
            return

        print("\nArray created successfully:")
        print("\n", self.array)

        self.IndexingSlicing()

    def IndexingSlicing(self):

        if not self.CheckArray():
            return

        while True:

            print("\nChoose an operation:")
            print("\n1. Indexing")
            print("\n2. Slicing")
            print("\n3. Go Back")

            choice = self.GetInteger(
                "\nEnter your choice: "
            )

            if choice == 1:

                if self.array.ndim == 1:

                    index = self.GetInteger(
                        "Enter the index: "
                    )

                    try:
                        print(
                            "\nElement at index:",
                            self.array[index]
                        )
                    except IndexError:
                        print("Index out of range.")

                elif self.array.ndim == 2:

                    row = self.GetInteger(
                        "Enter the row index: "
                    )

                    column = self.GetInteger(
                        "Enter the column index: "
                    )

                    try:
                        print(
                            "\nElement at given index:",
                            self.array[row, column]
                        )
                    except IndexError:
                        print("Index out of range.")

                elif self.array.ndim == 3:

                    layer = self.GetInteger(
                        "Enter the layer index: "
                    )

                    row = self.GetInteger(
                        "Enter the row index: "
                    )

                    column = self.GetInteger(
                        "Enter the column index: "
                    )

                    try:
                        print(
                            "\nElement at given index:",
                            self.array[layer, row, column]
                        )
                    except IndexError:
                        print("Index out of range.")

            elif choice == 2:

                if self.array.ndim == 1:

                    slicing = input(
                        "Enter the range (start:end): "
                    )

                    try:
                        parts = slicing.split(":")

                        start = int(parts[0]) if parts[0] else None
                        stop = int(parts[1]) if parts[1] else None

                        print(
                            "\nSliced Array:",
                            self.array[start:stop]
                        )

                    except:
                        print("Invalid slicing format.")

                elif self.array.ndim == 2:

                    row_range = input(
                        "Enter the row range (start:end): "
                    )

                    column_range = input(
                        "Enter the column range (start:end): "
                    )

                    try:
                        r = row_range.split(":")
                        c = column_range.split(":")

                        r_start = int(r[0]) if r[0] else None
                        r_end = int(r[1]) if r[1] else None

                        c_start = int(c[0]) if c[0] else None
                        c_end = int(c[1]) if c[1] else None

                        result = self.array[
                            r_start:r_end,
                            c_start:c_end
                        ]

                        print("\nSliced Array:", result)

                    except:
                        print("Invalid slicing format.")

                elif self.array.ndim == 3:

                    layer_range = input(
                        "Enter the layer range (start:end): "
                    )

                    row_range = input(
                        "Enter the row range (start:end): "
                    )

                    column_range = input(
                        "Enter the column range (start:end): "
                    )

                    try:
                        l = layer_range.split(":")
                        r = row_range.split(":")
                        c = column_range.split(":")

                        l_start = int(l[0]) if l[0] else None
                        l_end = int(l[1]) if l[1] else None

                        r_start = int(r[0]) if r[0] else None
                        r_end = int(r[1]) if r[1] else None

                        c_start = int(c[0]) if c[0] else None
                        c_end = int(c[1]) if c[1] else None

                        result = self.array[
                            l_start:l_end,
                            r_start:r_end,
                            c_start:c_end
                        ]

                        print("\nSliced Array:", result)

                    except:
                        print("Invalid slicing format.")

            elif choice == 3:
                break

            else:
                print("Invalid choice.")

    def MathematicalOperations(self):

        if not self.CheckArray():
            return

        print("\nMathematical Operations:")
        print("\nChoose a mathematical operation:")
        print("\n1. Addition")
        print("\n2. Subtraction")
        print("\n3. Multiplication")
        print("\n4. Division")
        print("\n5. Dot Product")
        print("\n6. Matrix Multiplication")

        choice = self.GetInteger(
            "\nEnter your choice: "
        )

        if choice in [1, 2, 3, 4]:

            total = self.array.size

            while True:

                values = input(
                    f"Enter the same-size array elements ({total} elements separated by space): "
                ).split()

                if len(values) != total:
                    print(f"Please enter exactly {total} elements.")
                    continue

                try:
                    values = [float(x) for x in values]

                    if all(x.is_integer() for x in values):
                        values = [int(x) for x in values]

                    second = np.array(values).reshape(
                        self.array.shape
                    )

                    break

                except ValueError:
                    print("Invalid input. Please enter numbers only.")

            print("\nOriginal Array:", self.array)
            print("\nSecond Array:")
            print(second)

            if choice == 1:

                result = self.array + second

                print("\nResult of Addition:", result)

            elif choice == 2:

                result = self.array - second

                print("\nResult of Subtraction:", result)

            elif choice == 3:

                result = self.array * second

                print("\nResult of Multiplication:", result)

            elif choice == 4:

                if np.any(second == 0):
                    print("\nDivision by zero is not allowed.")
                else:
                    result = self.array / second
                    print("\nResult of Division:", result)

        elif choice == 5:

            if self.array.ndim == 1:

                second = self.__GetSecondArray(
                    self.array.shape
                )

                print("\nOriginal Array:", self.array)
                print("\nSecond Array:")
                print(second)

                print(
                    "\nDot Product:",
                    np.dot(self.array, second)
                )

            elif self.array.ndim == 2:

                rows, columns = self.array.shape

                second = self.__GetSecondArray(
                    (columns, rows)
                )

                print("\nOriginal Array:", self.array)
                print("\nSecond Array:")
                print(second)

                print(
                    "\nDot Product:"
                )
                print(
                    np.dot(self.array, second)
                )

            else:
                print(
                    "Dot Product is supported only for 1D and 2D arrays."
                )

        elif choice == 6:

            if self.array.ndim != 2:

                print(
                    "Matrix multiplication requires 2D arrays."
                )

            else:

                rows, columns = self.array.shape

                second = self.__GetSecondArray(
                    (columns, rows)
                )

                print("\nOriginal Array:")
                print(self.array)

                print("\nSecond Array:")
                print(second)

                print("\nMatrix Multiplication:")
                print(
                    np.matmul(self.array, second)
                )

        else:
            print("Invalid choice.")

    def CombineSplit(self):

        if not self.CheckArray():
            return

        print("\nCombine or Split Arrays:")
        print("\nChoose an option:")
        print("\n1. Combine Arrays")
        print("\n2. Split Array")

        choice = self.GetInteger(
            "\nEnter your choice: "
        )

        if choice == 1:

            second = self.__GetSecondArray(
                self.array.shape
            )

            print("\nOriginal Array:")
            print(self.array)

            print("\nSecond Array:")
            print(second)

            print("\nChoose combining method:")
            print("\n1. Vertical Stack")
            print("\n2. Horizontal Stack")
            print("\n3. Concatenate")

            combine_choice = self.GetInteger(
                "\nEnter your choice: "
            )

            try:

                if combine_choice == 1:

                    result = np.vstack(
                        (self.array, second)
                    )

                    print(
                        "\nCombined Array (Vertical Stack):"
                    )
                    print(result)

                elif combine_choice == 2:

                    result = np.hstack(
                        (self.array, second)
                    )

                    print(
                        "\nCombined Array (Horizontal Stack):"
                    )
                    print(result)

                elif combine_choice == 3:

                    result = np.concatenate(
                        (self.array, second)
                    )

                    print(
                        "\nCombined Array (Concatenation):"
                    )
                    print(result)

                else:
                    print("Invalid choice.")

            except ValueError as error:
                print("Arrays cannot be combined:", error)

        elif choice == 2:

            number = self.GetInteger(
                "Enter the number of parts: "
            )

            if number <= 0:
                print("Number of parts must be greater than 0.")
                return

            try:

                result = np.array_split(
                    self.array,
                    number
                )

                print("\nSplit Arrays:")

                for i, part in enumerate(
                    result,
                    start=1
                ):
                    print(f"\nPart {i}:")
                    print(part)

            except ValueError as error:
                print("Unable to split array:", error)

        else:
            print("Invalid choice.")

    def SearchSortFilter(self):

        if not self.CheckArray():
            return

        print("\nSearch, Sort, and Filter:")
        print("\nChoose an option:")
        print("\n1. Search a value")
        print("\n2. Sort the array")
        print("\n3. Filter values")

        choice = self.GetInteger(
            "\nEnter your choice: "
        )

        if choice == 1:

            value = self.GetInteger(
                "Enter the value to search: "
            )

            indexes = np.where(
                self.array == value
            )

            if len(indexes[0]) == 0:

                print("\nValue not found.")

            else:

                print("\nIndex:", indexes)

        elif choice == 2:

            print(
                "\nOriginal Array:",
                self.array
            )

            print("\nChoose sorting order:")
            print("\n1. Ascending")
            print("\n2. Descending")

            sort_choice = self.GetInteger(
                "\nEnter your choice: "
            )

            result = np.sort(
                self.array,
                axis=-1
            )

            if sort_choice == 1:

                print("\nSorted Array:")
                print(result)

                print(
                    "\n(Sorting applied row-wise.)"
                )

            elif sort_choice == 2:

                result = np.flip(
                    result,
                    axis=-1
                )

                print("\nSorted Array:")
                print(result)

                print(
                    "\n(Sorting applied row-wise.)"
                )

            else:
                print("Invalid choice.")

        elif choice == 3:

            value = self.GetInteger(
                "Enter the comparison value: "
            )

            print("\nChoose filter condition:")
            print("\n1. Greater than")
            print("\n2. Less than")
            print("\n3. Equal to")
            print("\n4. Greater than or equal to")
            print("\n5. Less than or equal to")

            filter_choice = self.GetInteger(
                "\nEnter your choice: "
            )

            if filter_choice == 1:
                result = self.array[
                    self.array > value
                ]

            elif filter_choice == 2:
                result = self.array[
                    self.array < value
                ]

            elif filter_choice == 3:
                result = self.array[
                    self.array == value
                ]

            elif filter_choice == 4:
                result = self.array[
                    self.array >= value
                ]

            elif filter_choice == 5:
                result = self.array[
                    self.array <= value
                ]

            else:
                print("Invalid choice.")
                return

            print("\nFiltered Values:")
            print(result)

        else:
            print("Invalid choice.")

    def AggregatesStatistics(self):

        if not self.CheckArray():
            return

        print("\nAggregates and Statistics:")
        print("\nChoose an aggregate/statistical operation:")
        print("\n1. Sum")
        print("\n2. Mean")
        print("\n3. Median")
        print("\n4. Standard Deviation")
        print("\n5. Variance")
        print("\n6. Minimum")
        print("\n7. Maximum")
        print("\n8. Percentile")
        print("\n9. Correlation Coefficient")

        choice = self.GetInteger(
            "\nEnter your choice: "
        )

        print("\nOriginal Array:")
        print(self.array)

        if choice == 1:

            print(
                "\nSum of Array:",
                np.sum(self.array)
            )

        elif choice == 2:

            print(
                "\nMean of Array:",
                np.mean(self.array)
            )

        elif choice == 3:

            print(
                "\nMedian of Array:",
                np.median(self.array)
            )

        elif choice == 4:

            print(
                "\nStandard Deviation of Array:",
                np.std(self.array)
            )

        elif choice == 5:

            print(
                "\nVariance of Array:",
                np.var(self.array)
            )

        elif choice == 6:

            print(
                "\nMinimum of Array:",
                np.min(self.array)
            )

        elif choice == 7:

            print(
                "\nMaximum of Array:",
                np.max(self.array)
            )

        elif choice == 8:

            percentile = self.GetInteger(
                "Enter percentile (0-100): "
            )

            if 0 <= percentile <= 100:

                print(
                    f"\n{percentile}th Percentile of Array:",
                    np.percentile(
                        self.array,
                        percentile
                    )
                )

            else:
                print(
                    "Percentile must be between 0 and 100."
                )

        elif choice == 9:

            second = self.__GetSecondArray(
                self.array.shape
            )

            first_flat = self.array.reshape(-1)
            second_flat = second.reshape(-1)

            correlation = np.corrcoef(
                first_flat,
                second_flat
            )[0, 1]

            print(
                "\nCorrelation Coefficient:",
                correlation
            )

        else:
            print("Invalid choice.")

    def MainMenu(self):

        while True:

            print("\nWelcome to the NumPy Analyzer!")
            print("\nChoose an option:")
            print("\n1. Create a Numpy Array")
            print("\n2. Perform Mathematical Operations")
            print("\n3. Combine or Split Arrays")
            print("\n4. Search, Sort, or Filter Arrays")
            print("\n5. Compute Aggregates and Statistics ")
            print("\n6. Exit ")

            choice = self.GetInteger(
                "\nEnter your choice: "
            )

            if choice == 1:

                self.CreateArray()

            elif choice == 2:

                self.MathematicalOperations()

            elif choice == 3:

                self.CombineSplit()

            elif choice == 4:

                self.SearchSortFilter()

            elif choice == 5:

                self.AggregatesStatistics()

            elif choice == 6:

                print(
                    "\nThank you for using the NumPy Analyzer! Goodbye!"
                )
                break

            else:

                print(
                    "Invalid choice. Please try again."
                )
```

```python


analyzer = DataAnalytics()

analyzer.MainMenu()
```
