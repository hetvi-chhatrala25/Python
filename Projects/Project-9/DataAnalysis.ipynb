```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python



class SalesDataAnalyzer:

    def __init__(self):
        self.data = pd.DataFrame()
        self.current_plot = None
        self.all_plots = []
        sns.set_theme(style="whitegrid")
        print("\nSalesDataAnalyzer object created.")

    def __del__(self):
        print("\nSalesDataAnalyzer object destroyed.")

    def CheckData(self):
        if self.data.empty:
            print("\nNo dataset has been loaded yet.")
            return False
        return True

    def LoadData(self, file_path):
        try:
            self.data = pd.read_csv(file_path)
            self.data.columns = self.data.columns.astype(str).str.strip()
            print("\nDataset loaded successfully!")
            print("Rows:", len(self.data))
            print("Columns:", len(self.data.columns))
            return True
        except FileNotFoundError:
            print("\nError: File not found.")
            return False
        except Exception as e:
            print("\nError while loading dataset:", e)
            return False

    def FindDatetimeColumn(self):
        if self.data.empty:
            return None

        possible_names = [
            "datetime", "date_time", "timestamp", "date", "time",
            "DateTime", "Date", "Time"
        ]

        for column in self.data.columns:
            if column in possible_names:
                return column

        for column in self.data.columns:
            if str(column).lower().strip() in [
                "datetime", "date_time", "timestamp", "date", "time"
            ]:
                return column

        return None

    def ExploreData(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== EXPLORE DATA ==========")
            print("1. First 5 Rows")
            print("2. Last 5 Rows")
            print("3. Column Names")
            print("4. Data Types")
            print("5. Dataset Information")
            print("6. Statistical Description")
            print("7. Shape")
            print("8. Index")
            print("9. Unique Values")
            print("10. Missing Values")
            print("11. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":
                print(self.data.head())
            elif choice == "2":
                print(self.data.tail())
            elif choice == "3":
                print(self.data.columns.tolist())
            elif choice == "4":
                print(self.data.dtypes)
            elif choice == "5":
                self.data.info()
            elif choice == "6":
                print(self.data.describe(include="all"))
            elif choice == "7":
                print("Shape:", self.data.shape)
            elif choice == "8":
                print(self.data.index)
            elif choice == "9":
                for column in self.data.columns:
                    print("\n", column)
                    print(self.data[column].unique())
            elif choice == "10":
                print(self.data.isnull().sum())
            elif choice == "11":
                break
            else:
                print("\nInvalid choice.")

    def DatetimeOperations(self):
        if not self.CheckData():
            return

        print("\nAvailable Columns:", self.data.columns.tolist())
        column = input("Enter datetime/date/time column: ")

        if column not in self.data.columns:
            print("\nInvalid column.")
            return

        converted = pd.to_datetime(self.data[column], errors="coerce")

        while True:
            print("\n========== DATETIME OPERATIONS ==========")
            print("1. Convert Column")
            print("2. Create Date Column")
            print("3. Create Time Column")
            print("4. Create Hour Column")
            print("5. Create Day Column")
            print("6. Create Month Column")
            print("7. Create Weekday Column")
            print("8. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":
                self.data[column] = converted
                print("Column converted successfully.")
            elif choice == "2":
                self.data["Date"] = converted.dt.date
                print("Date column created.")
            elif choice == "3":
                self.data["Time"] = converted.dt.time
                print("Time column created.")
            elif choice == "4":
                self.data["Hour"] = converted.dt.hour
                print("Hour column created.")
            elif choice == "5":
                self.data["Day"] = converted.dt.day
                print("Day column created.")
            elif choice == "6":
                self.data["Month"] = converted.dt.month
                print("Month column created.")
            elif choice == "7":
                self.data["Weekday"] = converted.dt.day_name()
                print("Weekday column created.")
            elif choice == "8":
                break
            else:
                print("Invalid choice.")

    def CleanData(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== DATA CLEANING ==========")
            print("1. Display Missing Values")
            print("2. Fill Numeric Missing Values with Mean")
            print("3. Drop Rows with Missing Values")
            print("4. Remove Duplicate Rows")
            print("5. Convert Datetime Column")
            print("6. Create Date Column")
            print("7. Create Time Column")
            print("8. Create Hour Column")
            print("9. Create Day Column")
            print("10. Create Month Column")
            print("11. Create Weekday Column")
            print("12. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":
                print(self.data.isnull().sum())

            elif choice == "2":
                numeric_columns = self.data.select_dtypes(
                    include=np.number
                ).columns
                for column in numeric_columns:
                    self.data[column] = self.data[column].fillna(
                        self.data[column].mean()
                    )
                print("Numeric missing values filled with mean.")

            elif choice == "3":
                before = len(self.data)
                self.data.dropna(inplace=True)
                print("Rows removed:", before - len(self.data))

            elif choice == "4":
                before = len(self.data)
                self.data.drop_duplicates(inplace=True)
                print("Duplicate rows removed:", before - len(self.data))

            elif choice in ["5", "6", "7", "8", "9", "10", "11"]:
                column = self.FindDatetimeColumn()
                if column is None:
                    print("Datetime column not found.")
                    column = input("Enter datetime column manually: ")

                if column not in self.data.columns:
                    print("Invalid datetime column.")
                    continue

                converted = pd.to_datetime(
                    self.data[column], errors="coerce"
                )

                if choice == "5":
                    self.data[column] = converted
                    print("Datetime column converted.")
                elif choice == "6":
                    self.data["Date"] = converted.dt.date
                    print("Date column created.")
                elif choice == "7":
                    self.data["Time"] = converted.dt.time
                    print("Time column created.")
                elif choice == "8":
                    self.data["Hour"] = converted.dt.hour
                    print("Hour column created.")
                elif choice == "9":
                    self.data["Day"] = converted.dt.day
                    print("Day column created.")
                elif choice == "10":
                    self.data["Month"] = converted.dt.month
                    print("Month column created.")
                elif choice == "11":
                    self.data["Weekday"] = converted.dt.day_name()
                    print("Weekday column created.")

            elif choice == "12":
                break
            else:
                print("Invalid choice.")

    def NumpyOperations(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()

        if len(numeric_columns) == 0:
            print("No numeric columns found.")
            return

        print("\nNumeric Columns:", numeric_columns)
        column = input("Enter numeric column: ")

        if column not in numeric_columns:
            print("Invalid numeric column.")
            return

        array = self.data[column].dropna().to_numpy()

        if len(array) == 0:
            print("No numeric values available.")
            return

        print("\n========== NUMPY OPERATIONS ==========")
        print("Array:", array)
        print("Type:", type(array))
        print("Shape:", array.shape)
        print("Size:", array.size)
        print("First element:", array[0])
        print("First 5 elements:", array[:5])
        print("Last 5 elements:", array[-5:])
        print("Minimum:", np.min(array))
        print("Maximum:", np.max(array))
        print("Mean:", np.mean(array))
        print("Standard deviation:", np.std(array))
        print("Array + 10:", array + 10)
        print("Array - 10:", array - 10)
        print("Array * 2:", array * 2)
        print("Array / 2:", array / 2)

        first_array = array[:5]
        second_array = array[5:10]
        print("Combined array:", np.concatenate((first_array, second_array)))

        split_arrays = np.array_split(array, 2)
        print("Split arrays:")
        for index, part in enumerate(split_arrays, start=1):
            print("Part", index, ":", part)

    def MathematicalOperations(self):
        if not self.CheckData():
            return

        if "money" not in self.data.columns:
            print("'money' column not found.")
            return

        money = pd.to_numeric(self.data["money"], errors="coerce")
        self.data["Double_Money"] = money * 2
        self.data["Half_Money"] = money / 2
        self.data["Money_Plus_10"] = money + 10
        self.data["Money_Minus_10"] = money - 10
        self.data["Money_Squared"] = money ** 2

        print("\nMathematical operations completed.")
        print(self.data[
            [
                "money", "Double_Money", "Half_Money",
                "Money_Plus_10", "Money_Minus_10", "Money_Squared"
            ]
        ].head())

    def CombineData(self):
        if not self.CheckData():
            return

        file_path = input("Enter second CSV file path: ")

        try:
            other_dataframe = pd.read_csv(file_path)
            other_dataframe.columns = (
                other_dataframe.columns.astype(str).str.strip()
            )
        except Exception as e:
            print("Could not load second file:", e)
            return

        print("\n1. Concatenate")
        print("2. Merge")
        print("3. Join")
        choice = input("Enter your choice: ")

        try:
            if choice == "1":
                self.data = pd.concat(
                    [self.data, other_dataframe], ignore_index=True
                )
                print("DataFrames concatenated successfully.")

            elif choice in ["2", "3"]:
                common_columns = [
                    column for column in self.data.columns
                    if column in other_dataframe.columns
                ]

                if len(common_columns) == 0:
                    print("No common columns available.")
                    return

                print("Common columns:", common_columns)
                column = input("Enter common column: ")

                if column not in common_columns:
                    print("Invalid column.")
                    return

                if choice == "2":
                    self.data = pd.merge(
                        self.data,
                        other_dataframe,
                        on=column,
                        how="outer"
                    )
                    print("DataFrames merged successfully.")
                else:
                    first = self.data.set_index(column)
                    second = other_dataframe.set_index(column)
                    self.data = first.join(
                        second,
                        lsuffix="_left",
                        rsuffix="_right",
                        how="outer"
                    ).reset_index()
                    print("DataFrames joined successfully.")
            else:
                print("Invalid choice.")
                return

            print(self.data.head())

        except Exception as e:
            print("Error while combining data:", e)

    def SplitData(self):
        if not self.CheckData():
            return

        print("Available columns:", self.data.columns.tolist())
        column = input("Enter column to split data by: ")

        if column not in self.data.columns:
            print("Invalid column.")
            return

        for name, group in self.data.groupby(column, dropna=False):
            print("\nGroup:", name)
            print("Rows:", len(group))
            print(group.head())

    def SearchSortFilter(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== SEARCH / SORT / FILTER ==========")
            print("1. Search")
            print("2. Sort")
            print("3. Filter")
            print("4. Back")
            choice = input("Enter your choice: ")

            if choice == "1":
                column = input("Enter column to search: ")
                if column not in self.data.columns:
                    print("Invalid column.")
                    continue

                value = input("Enter search text: ")
                result = self.data[
                    self.data[column].astype(str).str.contains(
                        value, case=False, na=False
                    )
                ]
                print(result)

            elif choice == "2":
                column = input("Enter column to sort: ")
                if column not in self.data.columns:
                    print("Invalid column.")
                    continue

                order = input(
                    "Enter A for ascending or D for descending: "
                ).upper()

                if order not in ["A", "D"]:
                    print("Invalid order.")
                    continue

                result = self.data.sort_values(
                    by=column, ascending=(order == "A")
                )
                print(result.head(20))

            elif choice == "3":
                column = input("Enter column to filter: ")
                if column not in self.data.columns:
                    print("Invalid column.")
                    continue

                if pd.api.types.is_numeric_dtype(self.data[column]):
                    print("Operators: >, <, >=, <=, ==")
                    operator = input("Enter operator: ")
                    try:
                        value = float(input("Enter numeric value: "))
                        series = self.data[column]

                        if operator == ">":
                            result = self.data[series > value]
                        elif operator == "<":
                            result = self.data[series < value]
                        elif operator == ">=":
                            result = self.data[series >= value]
                        elif operator == "<=":
                            result = self.data[series <= value]
                        elif operator == "==":
                            result = self.data[series == value]
                        else:
                            print("Invalid operator.")
                            continue

                        print(result)
                    except ValueError:
                        print("Please enter a valid number.")
                else:
                    value = input("Enter text to filter: ")
                    result = self.data[
                        self.data[column].astype(str).str.contains(
                            value, case=False, na=False
                        )
                    ]
                    print(result)

            elif choice == "4":
                break
            else:
                print("Invalid choice.")

    def AggregateFunctions(self):
        if not self.CheckData():
            return

        print("Available columns:", self.data.columns.tolist())
        group_column = input("Enter GroupBy column: ")
        value_column = input("Enter value column: ")

        if group_column not in self.data.columns:
            print("Invalid GroupBy column.")
            return
        if value_column not in self.data.columns:
            print("Invalid value column.")
            return

        print("\n1. Sum")
        print("2. Mean")
        print("3. Count")
        print("4. Min")
        print("5. Max")
        choice = input("Enter choice: ")

        if choice not in ["1", "2", "3", "4", "5"]:
            print("Invalid choice.")
            return

        try:
            if choice == "3":
                result = self.data.groupby(group_column)[value_column].count()
            else:
                numeric_values = pd.to_numeric(
                    self.data[value_column], errors="coerce"
                )
                temporary = self.data.copy()
                temporary["_numeric_value"] = numeric_values
                temporary.dropna(
                    subset=["_numeric_value"], inplace=True
                )

                if temporary.empty:
                    print("This operation requires numeric values.")
                    return

                grouped = temporary.groupby(group_column)["_numeric_value"]

                if choice == "1":
                    result = grouped.sum()
                elif choice == "2":
                    result = grouped.mean()
                elif choice == "4":
                    result = grouped.min()
                else:
                    result = grouped.max()

            print("\nGroupBy Result:")
            print(result)

        except Exception as e:
            print("Cannot perform this operation:", e)

    def StatisticalAnalysis(self):
        if not self.CheckData():
            return

        numeric_data = self.data.select_dtypes(include=np.number)

        if numeric_data.empty:
            print("No numeric data available.")
            return

        print("\n========== STATISTICAL ANALYSIS ==========")
        print("\nDescriptive statistics:")
        print(numeric_data.describe())
        print("\nStandard deviation:")
        print(numeric_data.std())
        print("\nVariance:")
        print(numeric_data.var())
        print("\n25% quantile:")
        print(numeric_data.quantile(0.25))
        print("\n50% quantile:")
        print(numeric_data.quantile(0.50))
        print("\n75% quantile:")
        print(numeric_data.quantile(0.75))

    def CreatePivotTable(self):
        if not self.CheckData():
            return

        print("Available columns:", self.data.columns.tolist())
        index_column = input("Enter index column: ")
        value_column = input("Enter value column: ")

        if index_column not in self.data.columns:
            print("Invalid index column.")
            return
        if value_column not in self.data.columns:
            print("Invalid value column.")
            return

        try:
            temporary = self.data.copy()
            temporary[value_column] = pd.to_numeric(
                temporary[value_column], errors="coerce"
            )
            pivot = pd.pivot_table(
                temporary,
                index=index_column,
                values=value_column,
                aggfunc=["sum", "mean", "count"]
            )
            print("\n========== PIVOT TABLE ==========")
            print(pivot)
        except Exception as e:
            print("Error creating pivot table:", e)

    def GroupbyTransform(self):
        if not self.CheckData():
            return

        print("Available columns:", self.data.columns.tolist())
        group_column = input("Enter GroupBy column: ")
        value_column = input("Enter numeric value column: ")

        if group_column not in self.data.columns:
            print("Invalid group column.")
            return
        if value_column not in self.data.columns:
            print("Invalid value column.")
            return

        try:
            temporary = self.data.copy()
            temporary[value_column] = pd.to_numeric(
                temporary[value_column], errors="coerce"
            )

            grouped = temporary.groupby(group_column)[value_column].sum()
            print("\nGroupBy Sum:")
            print(grouped)

            temporary["Group_Mean"] = temporary.groupby(
                group_column
            )[value_column].transform("mean")

            print("\nTransform Mean:")
            print(temporary[[group_column, value_column, "Group_Mean"]].head(20))
            self.data["Group_Mean"] = temporary["Group_Mean"]

        except Exception as e:
            print("Error:", e)

    def ReindexData(self):
        if not self.CheckData():
            return

        self.data.reset_index(drop=True, inplace=True)
        print("Data re-indexed successfully.")
        print("New index:", self.data.index)

    def StoreVisualization(self, figure):
        self.current_plot = figure
        self.all_plots.append(figure)

    def PrepareNumericColumn(self, column):
        values = pd.to_numeric(self.data[column], errors="coerce")
        return values

    def MatplotlibBar(self):
        if not self.CheckData():
            return

        print("Available columns:", self.data.columns.tolist())
        x_column = input("Enter categorical column: ")
        y_column = input("Enter numeric column: ")

        if x_column not in self.data.columns or y_column not in self.data.columns:
            print("Invalid column.")
            return

        temporary = self.data[[x_column, y_column]].copy()
        temporary[y_column] = pd.to_numeric(
            temporary[y_column], errors="coerce"
        )
        temporary.dropna(inplace=True)

        if temporary.empty:
            print("No valid data available.")
            return

        grouped = temporary.groupby(x_column)[y_column].sum()
        grouped = grouped.sort_values(ascending=False).head(10)

        figure, axis = plt.subplots(
            figsize=(11, 6), constrained_layout=True
        )
        axis.bar(grouped.index.astype(str), grouped.values, label=y_column)
        axis.set_title(f"{y_column} by {x_column}")
        axis.set_xlabel(x_column)
        axis.set_ylabel(y_column)
        axis.tick_params(axis="x", rotation=45)
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibLine(self):
        if not self.CheckData():
            return

        datetime_column = self.FindDatetimeColumn()
        if datetime_column is None:
            print("Datetime column not found.")
            return

        if "money" not in self.data.columns:
            print("'money' column not found.")
            return

        temporary = self.data.copy()
        temporary[datetime_column] = pd.to_datetime(
            temporary[datetime_column], errors="coerce"
        )
        temporary["money"] = pd.to_numeric(
            temporary["money"], errors="coerce"
        )
        temporary.dropna(subset=[datetime_column, "money"], inplace=True)

        if temporary.empty:
            print("No valid datetime or sales data.")
            return

        daily_sales = temporary.groupby(
            temporary[datetime_column].dt.date
        )["money"].sum()

        figure, axis = plt.subplots(
            figsize=(12, 6), constrained_layout=True
        )
        axis.plot(
            daily_sales.index,
            daily_sales.values,
            marker="o",
            markersize=3,
            label="Sales"
        )
        axis.set_title("Daily Coffee Sales")
        axis.set_xlabel("Date")
        axis.set_ylabel("Sales")
        axis.tick_params(axis="x", rotation=45)
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibScatter(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()

        print("Numeric columns:", numeric_columns)
        x_column = input("Enter X-axis column: ")
        y_column = input("Enter Y-axis column: ")

        if x_column not in numeric_columns or y_column not in numeric_columns:
            print("Both columns must be numeric.")
            return

        temporary = self.data[[x_column, y_column]].copy().dropna()

        figure, axis = plt.subplots(
            figsize=(10, 6), constrained_layout=True
        )
        axis.scatter(
            temporary[x_column],
            temporary[y_column],
            label="Sales Data"
        )
        axis.set_title(f"{y_column} vs {x_column}")
        axis.set_xlabel(x_column)
        axis.set_ylabel(y_column)
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibPie(self):
        if not self.CheckData():
            return

        column = input("Enter categorical column: ")
        if column not in self.data.columns:
            print("Invalid column.")
            return

        values = self.data[column].value_counts().head(10)
        if values.empty:
            print("No data available.")
            return

        figure, axis = plt.subplots(
            figsize=(9, 9), constrained_layout=True
        )
        axis.pie(
            values.values,
            labels=values.index.astype(str),
            autopct="%1.1f%%"
        )
        axis.set_title(f"Distribution of {column}")
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibHistogram(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()
        print("Numeric columns:", numeric_columns)
        column = input("Enter numeric column: ")

        if column not in numeric_columns:
            print("Invalid numeric column.")
            return

        figure, axis = plt.subplots(
            figsize=(10, 6), constrained_layout=True
        )
        axis.hist(
            self.data[column].dropna(),
            bins=10,
            edgecolor="black",
            label=column
        )
        axis.set_title(f"Distribution of {column}")
        axis.set_xlabel(column)
        axis.set_ylabel("Frequency")
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibStackplot(self):
        if not self.CheckData():
            return

        if "coffee_name" not in self.data.columns or "money" not in self.data.columns:
            print("This chart requires 'coffee_name' and 'money' columns.")
            return

        datetime_column = self.FindDatetimeColumn()
        if datetime_column is None:
            print("Datetime column not found.")
            return

        temporary = self.data.copy()
        temporary[datetime_column] = pd.to_datetime(
            temporary[datetime_column], errors="coerce"
        )
        temporary["money"] = pd.to_numeric(
            temporary["money"], errors="coerce"
        )
        temporary.dropna(
            subset=[datetime_column, "money", "coffee_name"], inplace=True
        )

        temporary["Weekday"] = temporary[datetime_column].dt.day_name()
        pivot = pd.pivot_table(
            temporary,
            index="Weekday",
            columns="coffee_name",
            values="money",
            aggfunc="sum",
            fill_value=0
        )

        weekdays = [
            "Monday", "Tuesday", "Wednesday", "Thursday",
            "Friday", "Saturday", "Sunday"
        ]
        pivot = pivot.reindex(weekdays, fill_value=0)

        if pivot.empty or len(pivot.columns) == 0:
            print("No data available for stack plot.")
            return

        figure, axis = plt.subplots(
            figsize=(13, 7), constrained_layout=True
        )
        axis.stackplot(
            range(len(pivot)),
            *[pivot[column].values for column in pivot.columns],
            labels=pivot.columns
        )
        axis.set_title("Coffee Sales by Weekday")
        axis.set_xlabel("Weekday")
        axis.set_ylabel("Sales")
        axis.set_xticks(range(len(pivot)))
        axis.set_xticklabels(pivot.index, rotation=45)
        axis.legend(loc="upper left", bbox_to_anchor=(1, 1))
        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibSubplots(self):
        if not self.CheckData():
            return

        if "coffee_name" not in self.data.columns or "money" not in self.data.columns:
            print("This chart requires 'coffee_name' and 'money' columns.")
            return

        datetime_column = self.FindDatetimeColumn()
        if datetime_column is None:
            print("Datetime column not found.")
            return

        temporary = self.data.copy()
        temporary[datetime_column] = pd.to_datetime(
            temporary[datetime_column], errors="coerce"
        )
        temporary["money"] = pd.to_numeric(
            temporary["money"], errors="coerce"
        )
        temporary.dropna(
            subset=[datetime_column, "money", "coffee_name"], inplace=True
        )

        coffee_sales = temporary.groupby("coffee_name")["money"].sum()
        coffee_sales = coffee_sales.sort_values(
            ascending=False
        ).head(10)

        hourly_sales = temporary.groupby(
            temporary[datetime_column].dt.hour
        )["money"].sum()

        figure, axes = plt.subplots(
            1, 2, figsize=(15, 6), constrained_layout=True
        )

        axes[0].bar(
            coffee_sales.index.astype(str),
            coffee_sales.values,
            label="Sales"
        )
        axes[0].set_title("Top Coffee Sales")
        axes[0].set_xlabel("Coffee")
        axes[0].set_ylabel("Sales")
        axes[0].tick_params(axis="x", rotation=45)
        axes[0].legend()

        axes[1].plot(
            hourly_sales.index,
            hourly_sales.values,
            marker="o",
            label="Hourly Sales"
        )
        axes[1].set_title("Sales by Hour")
        axes[1].set_xlabel("Hour")
        axes[1].set_ylabel("Sales")
        axes[1].legend()

        self.StoreVisualization(figure)
        plt.show()

    def MatplotlibVisualization(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== MATPLOTLIB ==========")
            print("1. Bar Chart")
            print("2. Line Chart")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Subplots")
            print("8. Back")
            choice = input("Enter your choice: ")

            if choice == "1":
                self.MatplotlibBar()
            elif choice == "2":
                self.MatplotlibLine()
            elif choice == "3":
                self.MatplotlibScatter()
            elif choice == "4":
                self.MatplotlibPie()
            elif choice == "5":
                self.MatplotlibHistogram()
            elif choice == "6":
                self.MatplotlibStackplot()
            elif choice == "7":
                self.MatplotlibSubplots()
            elif choice == "8":
                break
            else:
                print("Invalid choice.")

    def SeabornBox(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()
        print("Numeric columns:", numeric_columns)
        column = input("Enter numeric column: ")

        if column not in numeric_columns:
            print("Invalid column.")
            return

        figure, axis = plt.subplots(
            figsize=(9, 6), constrained_layout=True
        )
        sns.boxplot(
            data=self.data,
            y=column,
            linewidth=2,
            fliersize=5,
            ax=axis
        )
        axis.set_title(f"Box Plot of {column}")
        axis.set_ylabel(column)
        self.StoreVisualization(figure)
        plt.show()

    def SeabornHeatmap(self):
        if not self.CheckData():
            return

        numeric_data = self.data.select_dtypes(include=np.number)
        if numeric_data.shape[1] < 2:
            print("At least two numeric columns are required.")
            return

        correlation = numeric_data.corr()

        figure, axis = plt.subplots(
            figsize=(11, 8), constrained_layout=True
        )
        sns.heatmap(
            correlation,
            annot=True,
            cmap="viridis",
            linewidths=1,
            linecolor="black",
            fmt=".2f",
            ax=axis
        )
        axis.set_title("Correlation Heatmap")
        self.StoreVisualization(figure)
        plt.show()

    def SeabornCount(self):
        if not self.CheckData():
            return

        column = input("Enter categorical column: ")
        if column not in self.data.columns:
            print("Invalid column.")
            return

        figure, axis = plt.subplots(
            figsize=(11, 6), constrained_layout=True
        )
        sns.countplot(data=self.data, x=column, ax=axis)
        axis.set_title(f"Count Plot of {column}")
        axis.tick_params(axis="x", rotation=45)
        self.StoreVisualization(figure)
        plt.show()

    def SeabornBar(self):
        if not self.CheckData():
            return

        print("\nAvailable columns:")
        print(self.data.columns.tolist())

        x_column = input("\nEnter categorical column (X-axis): ").strip()
        y_column = input("Enter numerical column (Y-axis): ").strip()

        # Accept column names without case or extra-space problems.
        column_map = {
            str(column).strip().lower(): column
            for column in self.data.columns
        }

        x_column = column_map.get(x_column.lower())
        y_column = column_map.get(y_column.lower())

        if x_column is None:
            print("\nInvalid X-axis column.")
            return

        if y_column is None:
            print("\nInvalid Y-axis column.")
            return

        # Convert the selected Y column into numeric values.
        # This also handles values such as Rs. 38.7, $38.7, or 1,000.
        plot_data = self.data[[x_column, y_column]].copy()

        plot_data[y_column] = (
            plot_data[y_column]
            .astype(str)
            .str.replace(",", "", regex=False)
            .str.replace("₹", "", regex=False)
            .str.replace("$", "", regex=False)
            .str.replace("Rs.", "", regex=False)
            .str.strip()
        )

        plot_data[y_column] = pd.to_numeric(
            plot_data[y_column],
            errors="coerce"
        )

        # Remove missing category and invalid numeric values.
        plot_data = plot_data.dropna(subset=[x_column, y_column])

        if plot_data.empty:
            print("\nNo valid data available.")
            print("Check that the X-axis column is categorical")
            print("and the Y-axis column contains numeric values.")
            print("\nY-axis data preview:")
            print(self.data[y_column].head())
            return

        # Group by the categorical column and calculate total sales.
        plot_data = (
            plot_data
            .groupby(x_column, as_index=False, dropna=False)[y_column]
            .sum()
            .sort_values(by=y_column, ascending=False)
            .head(10)
        )

        if plot_data.empty:
            print("\nNo grouped data available for the bar plot.")
            return

        figure, axis = plt.subplots(
            figsize=(12, 6),
            constrained_layout=True
        )

        sns.barplot(
            data=plot_data,
            x=x_column,
            y=y_column,
            estimator="sum",
            errorbar=None,
            ax=axis
        )

        axis.set_title(f"Total {y_column} by {x_column}")
        axis.set_xlabel(x_column)
        axis.set_ylabel(f"Total {y_column}")
        axis.tick_params(axis="x", rotation=45)

        # Keep labels readable without using plt.tight_layout().
        figure.canvas.draw()

        self.StoreVisualization(figure)
        plt.show()

    def SeabornHistogram(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()
        print("Numeric columns:", numeric_columns)
        column = input("Enter numeric column: ")

        if column not in numeric_columns:
            print("Invalid numeric column.")
            return

        figure, axis = plt.subplots(
            figsize=(10, 6), constrained_layout=True
        )
        sns.histplot(
            data=self.data,
            x=column,
            bins=10,
            kde=True,
            ax=axis
        )
        axis.set_title(f"Distribution of {column}")
        axis.set_xlabel(column)
        axis.set_ylabel("Frequency")
        self.StoreVisualization(figure)
        plt.show()

    def SeabornLine(self):
        if not self.CheckData():
            return

        datetime_column = self.FindDatetimeColumn()
        if datetime_column is None or "money" not in self.data.columns:
            print("Datetime column or 'money' column not found.")
            return

        temporary = self.data.copy()
        temporary[datetime_column] = pd.to_datetime(
            temporary[datetime_column], errors="coerce"
        )
        temporary["money"] = pd.to_numeric(
            temporary["money"], errors="coerce"
        )
        temporary.dropna(subset=[datetime_column, "money"], inplace=True)

        daily_sales = temporary.groupby(
            temporary[datetime_column].dt.date
        )["money"].sum().reset_index()
        daily_sales.columns = ["Date", "money"]

        figure, axis = plt.subplots(
            figsize=(12, 6), constrained_layout=True
        )
        sns.lineplot(
            data=daily_sales,
            x="Date",
            y="money",
            marker="o",
            label="Sales",
            ax=axis
        )
        axis.set_title("Daily Coffee Sales")
        axis.set_xlabel("Date")
        axis.set_ylabel("Sales")
        axis.tick_params(axis="x", rotation=45)
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def SeabornScatter(self):
        if not self.CheckData():
            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns.tolist()
        print("Numeric columns:", numeric_columns)

        x_column = input("Enter X-axis column: ")
        y_column = input("Enter Y-axis column: ")

        if x_column not in numeric_columns or y_column not in numeric_columns:
            print("Both columns must be numeric.")
            return

        temporary = self.data[[x_column, y_column]].dropna()

        figure, axis = plt.subplots(
            figsize=(10, 6), constrained_layout=True
        )
        sns.scatterplot(
            data=temporary,
            x=x_column,
            y=y_column,
            label="Sales Data",
            ax=axis
        )
        axis.set_title(f"{y_column} vs {x_column}")
        axis.set_xlabel(x_column)
        axis.set_ylabel(y_column)
        axis.legend()
        self.StoreVisualization(figure)
        plt.show()

    def SeabornVisualization(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== SEABORN ==========")
            print("1. Box Plot")
            print("2. Heatmap")
            print("3. Count Plot")
            print("4. Bar Plot")
            print("5. Histogram")
            print("6. Line Plot")
            print("7. Scatter Plot")
            print("8. Back")
            choice = input("Enter your choice: ")

            if choice == "1":
                self.SeabornBox()
            elif choice == "2":
                self.SeabornHeatmap()
            elif choice == "3":
                self.SeabornCount()
            elif choice == "4":
                self.SeabornBar()
            elif choice == "5":
                self.SeabornHistogram()
            elif choice == "6":
                self.SeabornLine()
            elif choice == "7":
                self.SeabornScatter()
            elif choice == "8":
                break
            else:
                print("Invalid choice.")

    def VisualizeData(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== DATA VISUALIZATION ==========")
            print("1. Matplotlib")
            print("2. Seaborn")
            print("3. Back")
            choice = input("Enter your choice: ")

            if choice == "1":
                self.MatplotlibVisualization()
            elif choice == "2":
                self.SeabornVisualization()
            elif choice == "3":
                break
            else:
                print("Invalid choice.")

    def SaveVisualization(self):
        if len(self.all_plots) == 0:
            print("\nNo visualization available to save.")
            return

        folder_name = input(
            "Enter folder name (example: coffee_visualizations): "
        ).strip()

        if folder_name == "":
            print("Folder name cannot be empty.")
            return

        try:
            os.makedirs(folder_name, exist_ok=True)

            for index, figure in enumerate(self.all_plots, start=1):
                file_name = os.path.join(
                    folder_name, f"coffee_sales_{index}.png"
                )
                figure.savefig(
                    file_name,
                    dpi=300,
                    bbox_inches="tight"
                )

            print(
                f"\nAll {len(self.all_plots)} visualizations "
                "saved successfully."
            )
            print("Folder:", folder_name)

        except Exception as e:
            print("Error saving visualizations:", e)

    def ExportData(self):
        if not self.CheckData():
            return

        file_name = input(
            "Enter CSV file name (example: analyzed_sales.csv): "
        ).strip()

        if file_name == "":
            print("File name cannot be empty.")
            return

        try:
            self.data.to_csv(file_name, index=False)
            print("Analysis data exported successfully.")
        except Exception as e:
            print("Error exporting data:", e)

    def Conclusion(self):
        if not self.CheckData():
            return

        print("\n========== CONCLUSION ==========")
        print("Total Transactions:", len(self.data))

        if "money" in self.data.columns:
            money = pd.to_numeric(self.data["money"], errors="coerce")
            print("Total Sales:", money.sum())
            print("Average Sale:", money.mean())
            print("Maximum Sale:", money.max())
            print("Minimum Sale:", money.min())

        if "coffee_name" in self.data.columns:
            coffee_values = self.data["coffee_name"].dropna()
            if not coffee_values.empty:
                print("Most Sold Coffee:", coffee_values.value_counts().idxmax())

        if "cash_type" in self.data.columns:
            payment_values = self.data["cash_type"].dropna()
            if not payment_values.empty:
                print(
                    "Most Used Payment Method:",
                    payment_values.value_counts().idxmax()
                )

        print("\nAnalysis completed successfully.")

    def DataframeOperations(self):
        if not self.CheckData():
            return

        while True:
            print("\n========== DATAFRAME OPERATIONS ==========")
            print("1. NumPy Operations")
            print("2. Mathematical Operations")
            print("3. Combine Data")
            print("4. Split Data")
            print("5. Search, Sort and Filter")
            print("6. Aggregate Functions")
            print("7. Create Pivot Table")
            print("8. GroupBy and Transform")
            print("9. Re-index Data")
            print("10. Display Current Data")
            print("11. Datetime Operations")
            print("12. Back")
            choice = input("Enter your choice: ")

            if choice == "1":
                self.NumpyOperations()
            elif choice == "2":
                self.MathematicalOperations()
            elif choice == "3":
                self.CombineData()
            elif choice == "4":
                self.SplitData()
            elif choice == "5":
                self.SearchSortFilter()
            elif choice == "6":
                self.AggregateFunctions()
            elif choice == "7":
                self.CreatePivotTable()
            elif choice == "8":
                self.GroupbyTransform()
            elif choice == "9":
                self.ReindexData()
            elif choice == "10":
                print(self.data.head(20))
            elif choice == "11":
                self.DatetimeOperations()
            elif choice == "12":
                break
            else:
                print("Invalid choice.")

    def Run(self):
        while True:
            print("\n" + "=" * 60)
            print("       SALES DATA ANALYSIS AND VISUALIZATION TOOL")
            print("=" * 60)
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. DataFrame Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Export Analysis Data")
            print("9. Conclusion / Summary")
            print("10. Exit")

            choice = input("\nEnter your choice: ")

            if choice == "1":
                file_path = input("Enter CSV file path: ")
                self.LoadData(file_path)
            elif choice == "2":
                self.ExploreData()
            elif choice == "3":
                self.DataframeOperations()
            elif choice == "4":
                self.CleanData()
            elif choice == "5":
                self.StatisticalAnalysis()
            elif choice == "6":
                self.VisualizeData()
            elif choice == "7":
                self.SaveVisualization()
            elif choice == "8":
                self.ExportData()
            elif choice == "9":
                self.Conclusion()
            elif choice == "10":
                print("\nThank you for using Sales Data Analysis Tool!")
                break
            else:
                print("\nInvalid choice. Please try again.")


```
