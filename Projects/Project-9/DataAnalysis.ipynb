```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python



class SalesDataAnalyzer:

    

    def __init__(self):

        self.data = pd.DataFrame()
        self.current_plot = None
        self.all_plots = []

        sns.set_theme(style="whitegrid")

        print("\nSalesDataAnalyzer object created.")

    

    def __del__(self):

        print("\nSalesDataAnalyzer object destroyed.")

    

    def CheckData(self):

        if self.data.empty:

            print("\nNo dataset has been loaded yet.")

            return False

        return True

    

    def FindDatetimeColumn(self):

        if self.data.empty:

            return None

        self.data.columns = self.data.columns.str.strip()

        possible_columns = [
            "datetime",
            "date",
            "date_time",
            "timestamp",
            "DateTime",
            "Date",
            "Time"
        ]

        for column in self.data.columns:

            if column in possible_columns:

                return column

        for column in self.data.columns:

            column_lower = column.lower().strip()

            if column_lower in [
                "datetime",
                "date",
                "date_time",
                "timestamp"
            ]:

                return column

        return None

    
    def LoadData(self, file_path):

        try:

            self.data = pd.read_csv(file_path)

            self.data.columns = self.data.columns.str.strip()

            print("\nDataset loaded successfully!")

            return True

        except FileNotFoundError:

            print("\nError: File not found.")

            return False

        except Exception as e:

            print("\nError while loading dataset:")

            print(e)

            return False

    

    def ExploreData(self):

        if not self.CheckData():

            return

        while True:

            print("\n========== EXPLORE DATA ==========")

            print("1. First 5 Rows")
            print("2. Last 5 Rows")
            print("3. Column Names")
            print("4. Data Types")
            print("5. Dataset Information")
            print("6. Statistical Description")
            print("7. Shape")
            print("8. Index")
            print("9. Unique Values")
            print("10. Missing Values")
            print("11. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":

                print("\nFirst 5 Rows:")

                print(self.data.head())

            elif choice == "2":

                print("\nLast 5 Rows:")

                print(self.data.tail())

            elif choice == "3":

                print("\nColumn Names:")

                print(self.data.columns.tolist())

            elif choice == "4":

                print("\nData Types:")

                print(self.data.dtypes)

            elif choice == "5":

                print("\nDataset Information:")

                self.data.info()

            elif choice == "6":

                print("\nStatistical Description:")

                print(self.data.describe(include="all"))

            elif choice == "7":

                print("\nDataset Shape:")

                print(self.data.shape)

            elif choice == "8":

                print("\nDataset Index:")

                print(self.data.index)

            elif choice == "9":

                print("\nUnique Values:")

                for column in self.data.columns:

                    print("\n", column)

                    print(self.data[column].unique())

            elif choice == "10":

                print("\nMissing Values:")

                print(self.data.isnull().sum())

            elif choice == "11":

                break

            else:

                print("\nInvalid choice.")

    

    def DatetimeOperations(self):

        if not self.CheckData():

            return

        datetime_column = self.FindDatetimeColumn()

        print("\nAvailable Columns:")

        print(self.data.columns.tolist())

        if datetime_column is None:

            print("\nDatetime column not found.")

            column_name = input(
                "\nEnter the datetime column name manually: "
            )

            if column_name not in self.data.columns:

                print("\nInvalid column name.")

                return

            datetime_column = column_name

        print("\nDatetime Column Found:")

        print(datetime_column)

        while True:

            print("\n========== DATETIME OPERATIONS ==========")

            print("1. Convert Datetime Column")
            print("2. Create Date Column")
            print("3. Create Time Column")
            print("4. Create Hour Column")
            print("5. Create Day Column")
            print("6. Create Month Column")
            print("7. Create Weekday Column")
            print("8. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                print(
                    "\nDatetime column converted successfully."
                )

            elif choice == "2":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Date"] = (
                    self.data[datetime_column].dt.date
                )

                print("\nDate column created successfully.")

            elif choice == "3":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Time"] = (
                    self.data[datetime_column].dt.time
                )

                print("\nTime column created successfully.")

            elif choice == "4":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Hour"] = (
                    self.data[datetime_column].dt.hour
                )

                print("\nHour column created successfully.")

            elif choice == "5":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Day"] = (
                    self.data[datetime_column].dt.day
                )

                print("\nDay column created successfully.")

            elif choice == "6":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Month"] = (
                    self.data[datetime_column].dt.month
                )

                print("\nMonth column created successfully.")

            elif choice == "7":

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                self.data["Weekday"] = (
                    self.data[datetime_column].dt.day_name()
                )

                print("\nWeekday column created successfully.")

            elif choice == "8":

                break

            else:

                print("\nInvalid choice.")

    

    def CleanData(self):

        if not self.CheckData():

            return

        while True:

            print("\n========== DATA CLEANING ==========")

            print("1. Display Missing Values")
            print("2. Fill Numeric Missing Values with Mean")
            print("3. Drop Rows with Missing Values")
            print("4. Remove Duplicate Rows")
            print("5. Convert Datetime Column")
            print("6. Create Date Column")
            print("7. Create Time Column")
            print("8. Create Hour Column")
            print("9. Create Day Column")
            print("10. Create Month Column")
            print("11. Create Weekday Column")
            print("12. Back")

            choice = input("\nEnter your choice: ")

            if choice == "1":

                print("\nMissing Values:")

                print(self.data.isnull().sum())

            elif choice == "2":

                numeric_columns = self.data.select_dtypes(
                    include=np.number
                ).columns

                for column in numeric_columns:

                    self.data[column] = self.data[column].fillna(
                        self.data[column].mean()
                    )

                print(
                    "\nNumeric missing values filled with mean."
                )

            elif choice == "3":

                before = len(self.data)

                self.data.dropna(inplace=True)

                after = len(self.data)

                print(
                    "\nRows removed:",
                    before - after
                )

            elif choice == "4":

                before = len(self.data)

                self.data.drop_duplicates(inplace=True)

                after = len(self.data)

                print(
                    "\nDuplicate rows removed:",
                    before - after
                )

            elif choice in [
                "5",
                "6",
                "7",
                "8",
                "9",
                "10",
                "11"
            ]:

                datetime_column = self.FindDatetimeColumn()

                if datetime_column is None:

                    print("\nDatetime column not found.")

                    continue

                self.data[datetime_column] = pd.to_datetime(
                    self.data[datetime_column],
                    errors="coerce"
                )

                if choice == "5":

                    print("\nDatetime column converted.")

                elif choice == "6":

                    self.data["Date"] = (
                        self.data[datetime_column].dt.date
                    )

                    print("\nDate column created.")

                elif choice == "7":

                    self.data["Time"] = (
                        self.data[datetime_column].dt.time
                    )

                    print("\nTime column created.")

                elif choice == "8":

                    self.data["Hour"] = (
                        self.data[datetime_column].dt.hour
                    )

                    print("\nHour column created.")

                elif choice == "9":

                    self.data["Day"] = (
                        self.data[datetime_column].dt.day
                    )

                    print("\nDay column created.")

                elif choice == "10":

                    self.data["Month"] = (
                        self.data[datetime_column].dt.month
                    )

                    print("\nMonth column created.")

                elif choice == "11":

                    self.data["Weekday"] = (
                        self.data[datetime_column].dt.day_name()
                    )

                    print("\nWeekday column created.")

            elif choice == "12":

                break

            else:

                print("\nInvalid choice.")

    

    def NumpyOperations(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        if len(numeric_columns) == 0:

            print("\nNo numeric columns found.")

            return

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        column = input(
            "\nEnter numeric column for NumPy operations: "
        )

        if column not in numeric_columns:

            print("\nInvalid numeric column.")

            return

        array = self.data[column].dropna().to_numpy()

        print("\n========== NUMPY OPERATIONS ==========")

        print("\nNumPy Array:")

        print(array)

        print("\nArray Type:")

        print(type(array))

        print("\nShape:")

        print(array.shape)

        print("\nSize:")

        print(array.size)

        if len(array) > 0:

            print("\nFirst Element:")

            print(array[0])

        if len(array) > 1:

            print("\nSecond Element:")

            print(array[1])

        print("\nFirst 5 Elements:")

        print(array[:5])

        print("\nLast 5 Elements:")

        print(array[-5:])

        print("\nFirst 10 Elements:")

        print(array[:10])

        print("\nMinimum:")

        print(np.min(array))

        print("\nMaximum:")

        print(np.max(array))

        print("\nMean:")

        print(np.mean(array))

        print("\nStandard Deviation:")

        print(np.std(array))

        print("\nElement-wise Addition (+10):")

        print(array + 10)

        print("\nElement-wise Subtraction (-10):")

        print(array - 10)

        print("\nElement-wise Multiplication (*2):")

        print(array * 2)

        print("\nElement-wise Division (/2):")

        print(array / 2)

        first_array = array[:5]

        second_array = array[5:10]

        combined_array = np.concatenate(
            (first_array, second_array)
        )

        print("\nCombined Array:")

        print(combined_array)

        if len(array) >= 2:

            split_arrays = np.array_split(array, 2)

            print("\nSplit Arrays:")

            for i, part in enumerate(
                split_arrays,
                start=1
            ):

                print("\nPart", i)

                print(part)

   

    def MathematicalOperations(self):

        if not self.CheckData():

            return

        if "money" not in self.data.columns:

            print("\n'money' column not found.")

            return

        self.data["money"] = pd.to_numeric(
            self.data["money"],
            errors="coerce"
        )

        self.data["Double_Money"] = (
            self.data["money"] * 2
        )

        self.data["Half_Money"] = (
            self.data["money"] / 2
        )

        self.data["Money_Plus_10"] = (
            self.data["money"] + 10
        )

        self.data["Money_Minus_10"] = (
            self.data["money"] - 10
        )

        self.data["Money_Squared"] = (
            self.data["money"] ** 2
        )

        print("\nMathematical operations completed.")

        print(
            self.data[
                [
                    "money",
                    "Double_Money",
                    "Half_Money",
                    "Money_Plus_10",
                    "Money_Minus_10",
                    "Money_Squared"
                ]
            ].head()
        )

    

    def CombineData(self, other_dataframe=None):

        if not self.CheckData():

            return

        try:

            if other_dataframe is None:

                file_path = input(
                    "\nEnter second CSV file path: "
                )

                other_dataframe = pd.read_csv(file_path)

            print("\nSecond DataFrame:")

            print(other_dataframe.head())

            print("\n1. Concatenate")
            print("2. Merge")
            print("3. Join")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                self.data = pd.concat(
                    [self.data, other_dataframe],
                    ignore_index=True
                )

                print(
                    "\nDataFrames concatenated successfully."
                )

            elif choice == "2":

                common_columns = list(
                    set(self.data.columns)
                    &
                    set(other_dataframe.columns)
                )

                if len(common_columns) == 0:

                    print(
                        "\nNo common columns available for merge."
                    )

                    return

                print("\nCommon Columns:")

                print(common_columns)

                column = input(
                    "\nEnter column for merge: "
                )

                if column not in common_columns:

                    print("\nInvalid column.")

                    return

                self.data = pd.merge(
                    self.data,
                    other_dataframe,
                    on=column,
                    how="outer"
                )

                print(
                    "\nDataFrames merged successfully."
                )

            elif choice == "3":

                common_columns = list(
                    set(self.data.columns)
                    &
                    set(other_dataframe.columns)
                )

                if len(common_columns) == 0:

                    print(
                        "\nNo common columns available for join."
                    )

                    return

                print("\nCommon Columns:")

                print(common_columns)

                column = input(
                    "\nEnter column to use as index: "
                )

                if column not in common_columns:

                    print("\nInvalid column.")

                    return

                first = self.data.set_index(column)

                second = other_dataframe.set_index(column)

                self.data = first.join(
                    second,
                    lsuffix="_left",
                    rsuffix="_right",
                    how="outer"
                ).reset_index()

                print(
                    "\nDataFrames joined successfully."
                )

            else:

                print("\nInvalid choice.")

                return

            print("\nUpdated Data:")

            print(self.data.head())

        except FileNotFoundError:

            print("\nSecond file not found.")

        except Exception as e:

            print("\nError while combining data:")

            print(e)

    
    def SplitData(self):

        if not self.CheckData():

            return

        print("\nAvailable Columns:")

        print(self.data.columns.tolist())

        column = input(
            "\nEnter column to split data by: "
        )

        if column not in self.data.columns:

            print("\nInvalid column.")

            return

        groups = self.data.groupby(column)

        print("\nGroups Found:")

        for name, group in groups:

            print("\nGroup:", name)

            print("Rows:", len(group))

            print(group.head())

    

    def SearchSortFilter(self):

        if not self.CheckData():

            return

        while True:

            print(
                "\n========== SEARCH / SORT / FILTER =========="
            )

            print("1. Search")
            print("2. Sort")
            print("3. Filter")
            print("4. Back")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                column = input(
                    "\nEnter column to search: "
                )

                if column not in self.data.columns:

                    print("\nInvalid column.")

                    continue

                value = input(
                    "\nEnter search value: "
                )

                result = self.data[
                    self.data[column]
                    .astype(str)
                    .str.contains(
                        value,
                        case=False,
                        na=False
                    )
                ]

                print("\nSearch Results:")

                print(result)

            elif choice == "2":

                column = input(
                    "\nEnter column to sort: "
                )

                if column not in self.data.columns:

                    print("\nInvalid column.")

                    continue

                order = input(
                    "\nEnter A for Ascending or D for Descending: "
                ).upper()

                if order == "A":

                    result = self.data.sort_values(
                        by=column,
                        ascending=True
                    )

                elif order == "D":

                    result = self.data.sort_values(
                        by=column,
                        ascending=False
                    )

                else:

                    print("\nInvalid order.")

                    continue

                print("\nSorted Data:")

                print(result.head(20))

            elif choice == "3":

                column = input(
                    "\nEnter column to filter: "
                )

                if column not in self.data.columns:

                    print("\nInvalid column.")

                    continue

                if pd.api.types.is_numeric_dtype(
                    self.data[column]
                ):

                    print("\nOperators:")

                    print(">")

                    print("<")

                    print(">=")

                    print("<=")

                    print("==")

                    operator = input(
                        "\nEnter operator: "
                    )

                    try:

                        value = float(
                            input(
                                "\nEnter numeric value: "
                            )
                        )

                        if operator == ">":

                            result = self.data[
                                self.data[column] > value
                            ]

                        elif operator == "<":

                            result = self.data[
                                self.data[column] < value
                            ]

                        elif operator == ">=":

                            result = self.data[
                                self.data[column] >= value
                            ]

                        elif operator == "<=":

                            result = self.data[
                                self.data[column] <= value
                            ]

                        elif operator == "==":

                            result = self.data[
                                self.data[column] == value
                            ]

                        else:

                            print("\nInvalid operator.")

                            continue

                        print("\nFiltered Data:")

                        print(result)

                    except ValueError:

                        print(
                            "\nPlease enter a valid number."
                        )

                else:

                    value = input(
                        "\nEnter value to filter: "
                    )

                    result = self.data[
                        self.data[column]
                        .astype(str)
                        .str.contains(
                            value,
                            case=False,
                            na=False
                        )
                    ]

                    print("\nFiltered Data:")

                    print(result)

            elif choice == "4":

                break

            else:

                print("\nInvalid choice.")


    def AggregateFunctions(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        if len(numeric_columns) == 0:

            print("\nNo numeric columns found.")

            return

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        column = input(
            "\nEnter numeric column: "
        )

        if column not in numeric_columns:

            print("\nInvalid column.")

            return

        print("\n========== AGGREGATE FUNCTIONS ==========")

        print("\nSum:")

        print(self.data[column].sum())

        print("\nMean:")

        print(self.data[column].mean())

        print("\nMinimum:")

        print(self.data[column].min())

        print("\nMaximum:")

        print(self.data[column].max())

        print("\nCount:")

        print(self.data[column].count())

        print("\nMedian:")

        print(self.data[column].median())

    

    def StatisticalAnalysis(self):

        if not self.CheckData():

            return

        numeric_data = self.data.select_dtypes(
            include=np.number
        )

        if numeric_data.empty:

            print("\nNo numeric data available.")

            return

        print(
            "\n========== STATISTICAL ANALYSIS =========="
        )

        print("\nDescriptive Statistics:")

        print(numeric_data.describe())

        print("\nStandard Deviation:")

        print(numeric_data.std())

        print("\nVariance:")

        print(numeric_data.var())

        print("\n25% Quantile:")

        print(numeric_data.quantile(0.25))

        print("\n50% Quantile:")

        print(numeric_data.quantile(0.50))

        print("\n75% Quantile:")

        print(numeric_data.quantile(0.75))

    

    def CreatePivotTable(self):

        if not self.CheckData():

            return

        print("\nAvailable Columns:")

        print(self.data.columns.tolist())

        index_column = input(
            "\nEnter index column: "
        )

        value_column = input(
            "\nEnter value column: "
        )

        if index_column not in self.data.columns:

            print("\nInvalid index column.")

            return

        if value_column not in self.data.columns:

            print("\nInvalid value column.")

            return

        try:

            pivot = pd.pivot_table(
                self.data,
                index=index_column,
                values=value_column,
                aggfunc=[
                    "sum",
                    "mean",
                    "count"
                ]
            )

            print("\n========== PIVOT TABLE ==========")

            print(pivot)

        except Exception as e:

            print("\nError creating pivot table:")

            print(e)


    def GroupbyTransform(self):

        if not self.CheckData():

            return

        print("\nAvailable Columns:")

        print(self.data.columns.tolist())

        group_column = input(
            "\nEnter column for GroupBy: "
        )

        value_column = input(
            "\nEnter numeric value column: "
        )

        if group_column not in self.data.columns:

            print("\nInvalid group column.")

            return

        if value_column not in self.data.columns:

            print("\nInvalid value column.")

            return

        try:

            grouped = (
                self.data
                .groupby(group_column)[value_column]
                .sum()
            )

            print("\nGroupBy Sum:")

            print(grouped)

            self.data["Group_Mean"] = (
                self.data
                .groupby(group_column)[value_column]
                .transform("mean")
            )

            print("\nTransform Mean Added:")

            print(
                self.data[
                    [
                        group_column,
                        value_column,
                        "Group_Mean"
                    ]
                ].head(20)
            )

        except Exception as e:

            print("\nError:")

            print(e)


    def ReindexData(self):

        if not self.CheckData():

            return

        self.data.reset_index(
            drop=True,
            inplace=True
        )

        print("\nData re-indexed successfully.")

        print("\nNew Index:")

        print(self.data.index)

    

    def StoreVisualization(self, fig):

        self.current_plot = fig

        self.all_plots.append(fig)

    
    def MatplotlibBar(self):

        if not self.CheckData():

            return

        print("\nAvailable Columns:")

        print(self.data.columns.tolist())

        x_column = input(
            "\nEnter categorical column: "
        )

        y_column = input(
            "\nEnter numeric column: "
        )

        if x_column not in self.data.columns:

            print("\nInvalid X column.")

            return

        if y_column not in self.data.columns:

            print("\nInvalid Y column.")

            return

        try:

            temp = (
                self.data
                .groupby(x_column)[y_column]
                .sum()
                .sort_values(ascending=False)
                .head(10)
            )

            fig, ax = plt.subplots(
                figsize=(10, 6)
            )

            ax.bar(
                temp.index.astype(str),
                temp.values,
                label=y_column
            )

            ax.set_title(
                f"{y_column} by {x_column}"
            )

            ax.set_xlabel(x_column)

            ax.set_ylabel(y_column)

            ax.tick_params(
                axis="x",
                rotation=45
            )

            ax.legend()

            plt.tight_layout()

            plt.show()

            self.StoreVisualization(fig)

        except Exception as e:

            print("\nError creating bar chart:")

            print(e)


    def MatplotlibLine(self):

        if not self.CheckData():

            return

        datetime_column = self.FindDatetimeColumn()

        if datetime_column is None:

            print("\nDatetime column not found.")

            return

        if "money" not in self.data.columns:

            print("\n'money' column not found.")

            return

        temp_data = self.data.copy()

        temp_data[datetime_column] = pd.to_datetime(
            temp_data[datetime_column],
            errors="coerce"
        )

        temp_data["money"] = pd.to_numeric(
            temp_data["money"],
            errors="coerce"
        )

        temp_data.dropna(
            subset=[
                datetime_column,
                "money"
            ],
            inplace=True
        )

        if temp_data.empty:

            print(
                "\nNo valid datetime or sales data."
            )

            return

        daily_sales = (
            temp_data
            .groupby(
                temp_data[datetime_column].dt.date
            )["money"]
            .sum()
        )

        fig, ax = plt.subplots(
            figsize=(12, 6)
        )

        ax.plot(
            daily_sales.index,
            daily_sales.values,
            marker="o",
            markersize=3,
            label="Sales"
        )

        ax.set_title(
            "Daily Coffee Sales"
        )

        ax.set_xlabel("Date")

        ax.set_ylabel("Sales")

        ax.legend()

        plt.xticks(rotation=45)

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def MatplotlibScatter(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        if len(numeric_columns) < 2:

            print(
                "\nAt least two numeric columns are required."
            )

            return

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        x_column = input(
            "\nEnter X-axis column: "
        )

        y_column = input(
            "\nEnter Y-axis column: "
        )

        if x_column not in numeric_columns:

            print("\nX column must be numeric.")

            return

        if y_column not in numeric_columns:

            print("\nY column must be numeric.")

            return

        plot_data = self.data[
            [x_column, y_column]
        ].dropna()

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        ax.scatter(
            plot_data[x_column],
            plot_data[y_column],
            label="Sales Data"
        )

        ax.set_title(
            f"{y_column} vs {x_column}"
        )

        ax.set_xlabel(x_column)

        ax.set_ylabel(y_column)

        ax.legend()

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def MatplotlibPie(self):

        if not self.CheckData():

            return

        column = input(
            "\nEnter categorical column: "
        )

        if column not in self.data.columns:

            print("\nInvalid column.")

            return

        values = (
            self.data[column]
            .value_counts()
            .head(10)
        )

        fig, ax = plt.subplots(
            figsize=(8, 8)
        )

        ax.pie(
            values.values,
            labels=values.index.astype(str),
            autopct="%1.1f%%"
        )

        ax.set_title(
            f"Distribution of {column}"
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def MatplotlibHistogram(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        column = input(
            "\nEnter numeric column: "
        )

        if column not in numeric_columns:

            print("\nInvalid numeric column.")

            return

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        ax.hist(
            self.data[column].dropna(),
            bins=10,
            edgecolor="black"
        )

        ax.set_title(
            f"Distribution of {column}"
        )

        ax.set_xlabel(column)

        ax.set_ylabel("Frequency")

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def MatplotlibStackplot(self):

        if not self.CheckData():

            return

        datetime_column = self.FindDatetimeColumn()

        if datetime_column is None:

            print("\nDatetime column not found.")

            return

        if "coffee_name" not in self.data.columns:

            print(
                "\n'coffee_name' column not found."
            )

            return

        if "money" not in self.data.columns:

            print(
                "\n'money' column not found."
            )

            return

        temp = self.data.copy()

        temp[datetime_column] = pd.to_datetime(
            temp[datetime_column],
            errors="coerce"
        )

        temp["money"] = pd.to_numeric(
            temp["money"],
            errors="coerce"
        )

        temp.dropna(
            subset=[
                datetime_column,
                "money"
            ],
            inplace=True
        )

        temp["Weekday"] = (
            temp[datetime_column]
            .dt.day_name()
        )

        pivot = pd.pivot_table(
            temp,
            index="Weekday",
            columns="coffee_name",
            values="money",
            aggfunc="sum",
            fill_value=0
        )

        weekdays = [
            "Monday",
            "Tuesday",
            "Wednesday",
            "Thursday",
            "Friday",
            "Saturday",
            "Sunday"
        ]

        pivot = pivot.reindex(
            weekdays,
            fill_value=0
        )

        fig, ax = plt.subplots(
            figsize=(12, 6)
        )

        ax.stackplot(
            range(len(pivot)),
            *[
                pivot[column].values
                for column in pivot.columns
            ],
            labels=pivot.columns
        )

        ax.set_title(
            "Coffee Sales by Weekday"
        )

        ax.set_xlabel("Weekday")

        ax.set_ylabel("Sales")

        ax.set_xticks(
            range(len(pivot))
        )

        ax.set_xticklabels(
            pivot.index,
            rotation=45
        )

        ax.legend(
            loc="upper left",
            bbox_to_anchor=(1, 1)
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def MatplotlibSubplots(self):

        if not self.CheckData():

            return

        if "coffee_name" not in self.data.columns:

            print(
                "\n'coffee_name' column not found."
            )

            return

        if "money" not in self.data.columns:

            print(
                "\n'money' column not found."
            )

            return

        datetime_column = self.FindDatetimeColumn()

        if datetime_column is None:

            print("\nDatetime column not found.")

            return

        temp = self.data.copy()

        temp[datetime_column] = pd.to_datetime(
            temp[datetime_column],
            errors="coerce"
        )

        temp["money"] = pd.to_numeric(
            temp["money"],
            errors="coerce"
        )

        temp.dropna(
            subset=[
                datetime_column,
                "money"
            ],
            inplace=True
        )

        coffee_sales = (
            temp
            .groupby("coffee_name")["money"]
            .sum()
            .sort_values(ascending=False)
            .head(10)
        )

        hourly_sales = (
            temp
            .groupby(
                temp[datetime_column].dt.hour
            )["money"]
            .sum()
        )

        fig, axes = plt.subplots(
            1,
            2,
            figsize=(15, 6)
        )

        axes[0].bar(
            coffee_sales.index.astype(str),
            coffee_sales.values
        )

        axes[0].set_title(
            "Top Coffee Sales"
        )

        axes[0].set_xlabel(
            "Coffee"
        )

        axes[0].set_ylabel(
            "Sales"
        )

        axes[0].tick_params(
            axis="x",
            rotation=45
        )

        axes[1].plot(
            hourly_sales.index,
            hourly_sales.values,
            marker="o",
            label="Hourly Sales"
        )

        axes[1].set_title(
            "Sales by Hour"
        )

        axes[1].set_xlabel(
            "Hour"
        )

        axes[1].set_ylabel(
            "Sales"
        )

        axes[1].legend()

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)

    def MatplotlibVisualization(self):

        if not self.CheckData():

            return

        while True:

            print(
                "\n========== MATPLOTLIB =========="
            )

            print("1. Bar Chart")
            print("2. Line Chart")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Subplots")
            print("8. Back")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                self.MatplotlibBar()

            elif choice == "2":

                self.MatplotlibLine()

            elif choice == "3":

                self.MatplotlibScatter()

            elif choice == "4":

                self.MatplotlibPie()

            elif choice == "5":

                self.MatplotlibHistogram()

            elif choice == "6":

                self.MatplotlibStackplot()

            elif choice == "7":

                self.MatplotlibSubplots()

            elif choice == "8":

                break

            else:

                print("\nInvalid choice.")


    def SeabornBox(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        column = input(
            "\nEnter numeric column: "
        )

        if column not in numeric_columns:

            print("\nInvalid column.")

            return

        fig, ax = plt.subplots(
            figsize=(8, 6)
        )

        sns.boxplot(
            data=self.data,
            y=column,
            linewidth=2,
            fliersize=5,
            ax=ax
        )

        ax.set_title(
            f"Box Plot of {column}"
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornHeatmap(self):

        if not self.CheckData():

            return

        numeric_data = self.data.select_dtypes(
            include=np.number
        )

        if numeric_data.shape[1] < 2:

            print(
                "\nAt least two numeric columns are required."
            )

            return

        correlation = numeric_data.corr()

        fig, ax = plt.subplots(
            figsize=(10, 7)
        )

        sns.heatmap(
            correlation,
            annot=True,
            cmap="viridis",
            linewidths=1,
            linecolor="black",
            fmt=".2f",
            ax=ax
        )

        ax.set_title(
            "Correlation Heatmap"
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornCount(self):

        if not self.CheckData():

            return

        column = input(
            "\nEnter categorical column: "
        )

        if column not in self.data.columns:

            print("\nInvalid column.")

            return

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        sns.countplot(
            data=self.data,
            x=column,
            ax=ax
        )

        ax.set_title(
            f"Count Plot of {column}"
        )

        ax.tick_params(
            axis="x",
            rotation=45
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornBar(self):
    
     if not self.CheckData():
        return

     try:
        print("\nAvailable Columns:")
        print(list(self.data.columns))

        x_column = input("\nEnter categorical column (X-axis): ")
        y_column = input("Enter numerical column (Y-axis): ")

        if x_column not in self.data.columns:
            print("\nInvalid X-axis column.")
            return

        if y_column not in self.data.columns:
            print("\nInvalid Y-axis column.")
            return

        plot_data = self.data[[x_column, y_column]].copy()

        plot_data[y_column] = pd.to_numeric(
            plot_data[y_column],
            errors="coerce"
        )

        plot_data.dropna(
            subset=[x_column, y_column],
            inplace=True
        )

        if plot_data.empty:
            print("\nNo valid data available.")
            return

        plot_data = (
            plot_data
            .groupby(x_column, as_index=False)[y_column]
            .sum()
        )

        fig, ax = plt.subplots(figsize=(12, 6))

        sns.barplot(
            data=plot_data,
            x=x_column,
            y=y_column,
            ax=ax,
            errorbar=None
        )

        ax.set_title(f"{x_column} by {y_column}")
        ax.set_xlabel(x_column)
        ax.set_ylabel(y_column)

        ax.tick_params(axis="x", rotation=45)

        fig.subplots_adjust(bottom=0.30)

        self.StoreVisualization(fig)

        plt.show()

     except Exception as e:
        print("\nError creating Seaborn bar chart:")
        print(e)  
        
        
    def SeabornHistogram(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        column = input(
            "\nEnter numeric column: "
        )

        if column not in numeric_columns:

            print("\nInvalid numeric column.")

            return

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        sns.histplot(
            data=self.data,
            x=column,
            bins=10,
            kde=True,
            ax=ax
        )

        ax.set_title(
            f"Distribution of {column}"
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornLine(self):

        if not self.CheckData():

            return

        datetime_column = self.FindDatetimeColumn()

        if datetime_column is None:

            print("\nDatetime column not found.")

            return

        if "money" not in self.data.columns:

            print("\n'money' column not found.")

            return

        temp = self.data.copy()

        temp[datetime_column] = pd.to_datetime(
            temp[datetime_column],
            errors="coerce"
        )

        temp["money"] = pd.to_numeric(
            temp["money"],
            errors="coerce"
        )

        temp.dropna(
            subset=[
                datetime_column,
                "money"
            ],
            inplace=True
        )

        daily_sales = (
            temp
            .groupby(
                temp[datetime_column].dt.date
            )["money"]
            .sum()
            .reset_index()
        )

        daily_sales.columns = [
            "Date",
            "money"
        ]

        fig, ax = plt.subplots(
            figsize=(12, 6)
        )

        sns.lineplot(
            data=daily_sales,
            x="Date",
            y="money",
            marker="o",
            ax=ax
        )

        ax.set_title(
            "Daily Coffee Sales"
        )

        ax.tick_params(
            axis="x",
            rotation=45
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornScatter(self):

        if not self.CheckData():

            return

        numeric_columns = self.data.select_dtypes(
            include=np.number
        ).columns

        if len(numeric_columns) < 2:

            print(
                "\nAt least two numeric columns are required."
            )

            return

        print("\nNumeric Columns:")

        print(numeric_columns.tolist())

        x_column = input(
            "\nEnter X-axis column: "
        )

        y_column = input(
            "\nEnter Y-axis column: "
        )

        if x_column not in numeric_columns:

            print("\nInvalid X column.")

            return

        if y_column not in numeric_columns:

            print("\nInvalid Y column.")

            return

        temp = self.data[
            [x_column, y_column]
        ].dropna()

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        sns.scatterplot(
            data=temp,
            x=x_column,
            y=y_column,
            ax=ax
        )

        ax.set_title(
            f"{y_column} vs {x_column}"
        )

        plt.tight_layout()

        plt.show()

        self.StoreVisualization(fig)


    def SeabornVisualization(self):

        if not self.CheckData():

            return

        while True:

            print(
                "\n========== SEABORN =========="
            )

            print("1. Box Plot")
            print("2. Heatmap")
            print("3. Count Plot")
            print("4. Bar Plot")
            print("5. Histogram")
            print("6. Line Plot")
            print("7. Scatter Plot")
            print("8. Back")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                self.SeabornBox()

            elif choice == "2":

                self.SeabornHeatmap()

            elif choice == "3":

                self.SeabornCount()

            elif choice == "4":

                self.SeabornBar()

            elif choice == "5":

                self.SeabornHistogram()

            elif choice == "6":

                self.SeabornLine()

            elif choice == "7":

                self.SeabornScatter()

            elif choice == "8":

                break

            else:

                print("\nInvalid choice.")


    def VisualizeData(self):

        if not self.CheckData():

            return

        while True:

            print(
                "\n========== DATA VISUALIZATION =========="
            )

            print("1. Matplotlib")
            print("2. Seaborn")
            print("3. Back")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                self.MatplotlibVisualization()

            elif choice == "2":

                self.SeabornVisualization()

            elif choice == "3":

                break

            else:

                print("\nInvalid choice.")


    def SaveVisualization(self):

        if len(self.all_plots) == 0:

            print(
                "\nNo visualization available to save."
            )

            return

        folder_name = input(
            "\nEnter folder name "
            "(example: coffee_visualizations): "
        )

        import os

        try:

            os.makedirs(
                folder_name,
                exist_ok=True
            )

            for index, plot in enumerate(
                self.all_plots,
                start=1
            ):

                file_name = os.path.join(
                    folder_name,
                    f"coffee_sales_{index}.png"
                )

                plot.savefig(
                    file_name,
                    dpi=300,
                    bbox_inches="tight"
                )

            print(
                f"\nAll {len(self.all_plots)} "
                "visualizations saved successfully."
            )

            print(
                f"Folder: {folder_name}"
            )

        except Exception as e:

            print(
                "\nError saving visualizations:"
            )

            print(e)

    def ExportData(self):

        if not self.CheckData():

            return

        file_name = input(
            "\nEnter CSV file name "
            "(example: analyzed_coffee_sales.csv): "
        )

        try:

            self.data.to_csv(
                file_name,
                index=False
            )

            print(
                "\nAnalysis data exported successfully."
            )

        except Exception as e:

            print(
                "\nError exporting data:"
            )

            print(e)


    def Conclusion(self):

        if not self.CheckData():

            return

        print("\n========== CONCLUSION ==========")

        print(
            "\nTotal Transactions:"
        )

        print(len(self.data))

        if "money" in self.data.columns:

            money = pd.to_numeric(
                self.data["money"],
                errors="coerce"
            )

            print(
                "\nTotal Sales:"
            )

            print(money.sum())

            print(
                "\nAverage Sale:"
            )

            print(money.mean())

            print(
                "\nMaximum Sale:"
            )

            print(money.max())

            print(
                "\nMinimum Sale:"
            )

            print(money.min())

        if "coffee_name" in self.data.columns:

            print(
                "\nMost Sold Coffee:"
            )

            print(
                self.data["coffee_name"]
                .value_counts()
                .idxmax()
            )

        if "cash_type" in self.data.columns:

            print(
                "\nMost Used Payment Method:"
            )

            print(
                self.data["cash_type"]
                .value_counts()
                .idxmax()
            )

        print(
            "\nAnalysis completed successfully."
        )


    def DataframeOperations(self):

        if not self.CheckData():

            return

        while True:

            print(
                "\n========== DATAFRAME OPERATIONS =========="
            )

            print("1. NumPy Operations")
            print("2. Mathematical Operations")
            print("3. Combine Data")
            print("4. Split Data")
            print("5. Search, Sort and Filter")
            print("6. Aggregate Functions")
            print("7. Create Pivot Table")
            print("8. GroupBy and Transform")
            print("9. Re-index Data")
            print("10. Display Current Data")
            print("11. Datetime Operations")
            print("12. Back")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                self.NumpyOperations()

            elif choice == "2":

                self.MathematicalOperations()

            elif choice == "3":

                self.CombineData()

            elif choice == "4":

                self.SplitData()

            elif choice == "5":

                self.SearchSortFilter()

            elif choice == "6":

                self.AggregateFunctions()

            elif choice == "7":

                self.CreatePivotTable()

            elif choice == "8":

                self.GroupbyTransform()

            elif choice == "9":

                self.ReindexData()

            elif choice == "10":

                print("\nCurrent Data:")

                print(self.data.head(20))

            elif choice == "11":

                self.DatetimeOperations()

            elif choice == "12":

                break

            else:

                print("\nInvalid choice.")

    def Run(self):

        while True:

            print("\n")

            print("=" * 55)

            print(
                "       SALES DATA ANALYSIS TOOL"
            )

            print("=" * 55)

            print("\n1. Load Dataset")
            print("2. Explore Data")
            print("3. DataFrame Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Export Analysis Data")
            print("9. Conclusion / Summary")
            print("10. Exit")

            choice = input(
                "\nEnter your choice: "
            )

            if choice == "1":

                file_path = input(
                    "\nEnter CSV file path: "
                )

                self.LoadData(file_path)

            elif choice == "2":

                self.ExploreData()

            elif choice == "3":

                self.DataframeOperations()

            elif choice == "4":

                self.CleanData()

            elif choice == "5":

                self.StatisticalAnalysis()

            elif choice == "6":

                self.VisualizeData()

            elif choice == "7":

                self.SaveVisualization()

            elif choice == "8":

                self.ExportData()

            elif choice == "9":

                self.Conclusion()

            elif choice == "10":

                print(
                    "\nThank you for using "
                    "Sales Data Analysis Tool!"
                )

                break

            else:

                print(
                    "\nInvalid choice. "
                    "Please try again."
                )




```
